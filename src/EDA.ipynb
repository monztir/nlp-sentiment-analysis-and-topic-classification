## Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import utility
import sys
import os

# Dodaj src folder u sys.path
sys.path.append(os.path.abspath("../src"))
```

```python
def load_data(path_imdb, path_20newsgroups): 
    data_imdb = utility.load_data_imdb(path_imdb) 
    data_20newsgroups = utility.load_data_20newsgroups(path_20newsgroups) 

    return data_imdb, data_20newsgroups
```

```python
path_imdb = "../data/raw/imdb/imdb.csv"
path_20newsgroups = "../data/raw/20_newsgroup"

data_imdb, data_20newsgroups = load_data(path_imdb, path_20newsgroups)
```

## **Section 1:** EDA - IMDB

```python
print("\nFirst few rows of IMDB dataset:")
print(data_imdb.head())

print("\nShape of IMDB dataset:")
print(data_imdb.shape)

print("\nData types in IMDB dataset:")
print(data_imdb.dtypes)

print("\nColumns in IMDB dataset:")
print(data_imdb.columns)

print("\nDescriptive statistics of IMDB dataset:")
print(data_imdb.describe())

print("\nMissing values in IMDB dataset:")
print(data_imdb.isnull().sum())

sentiment_counts = data_imdb['sentiment'].value_counts()

print("\nUnique values in 'sentiment' column of IMDB dataset:")
print(sentiment_counts)

plt.figure(figsize=(6, 4))
plt.bar(sentiment_counts.index, sentiment_counts.values, color=['blue', 'red'], alpha=0.6)
plt.title("IMDB dataset: Sentiment distribution")
plt.xlabel("Sentiment")
plt.ylabel("Count")
plt.show()

review_lengths = data_imdb["review"].apply(len)

print("Basic statistics of review length (in characters):")
print(review_lengths.describe())

plt.figure(figsize=(6, 4))
plt.hist(review_lengths, bins=50, color='blue', edgecolor='black')
plt.title("IMDB dataset: Histogram of review length")
plt.xlabel("Number of characters")
plt.ylabel("Frequency")
plt.show()
```

## **Section 2:** EDA - 20Newsgroups

```python
print("\nFirst few rows of 20Newsgroups dataset:")
print(data_20newsgroups.head())

print("\nShape of 20Newsgroups dataset:")
print(data_20newsgroups.shape)

print("\nData types in 20Newsgroups dataset:")
print(data_20newsgroups.dtypes)

print("\nColumns in 20Newsgroups dataset:")
print(data_20newsgroups.columns)

print("\nDescriptiove statistic of 20Newsgroups dataset:")
print(data_20newsgroups.describe())

print("\nMissing values in 20Newsgroups dataset:")
print(data_20newsgroups.isnull().sum())

print("Newsgroups:", data_20newsgroups['newsgroup'].unique())

newsgroups_counts = data_20newsgroups['newsgroup'].value_counts()

plt.figure(figsize=(10, 8))
plt.bar(newsgroups_counts.index, newsgroups_counts.values, color='blue', alpha=0.6)
plt.title("20Newsgroups Dataset: Newsgroup ditribution")
plt.xticks(rotation=45)
plt.xlabel("Newsgroup")
plt.ylabel("Count")
plt.show()
```
